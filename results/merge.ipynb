```python
import pandas as pd

# Read the CSV file
df_p2 = pd.read_csv('results_raw_p2.csv')

# Extract col1 and col2 from the 'Type' column
df_p2['query_modality'] = df_p2['Type'].str.split('-').str[0]  # First part (e.g., "Spec")
df_p2['document_modality'] = df_p2['Type'].str.split('-').str[-1].str.replace(':', '')  # Last part (e.g., "Spec")
df_p2['embedding_method'] = 'Text Model'  

# Drop original column
df_p2 = df_p2.drop(columns=['Type'])

# Rename
df_p2 = df_p2.replace('Text', 'Text (AltGosling)')
df_p2 = df_p2.replace('LLM', 'Text (AltGosling + LLM)')

# Reorder columns (new columns first, then original)
cols = ['query_modality', 'document_modality', 'embedding_method'] + list(df_p2.columns[:-3])
df_p2 = df_p2[cols]

print(df_p2)

# df_p2.to_csv('results_p2.csv')
```

```python
# Read part 1 from raw

df_p1_raw = pd.read_csv('results_p1.csv')
rename_dict = {col: col.replace("_", "=") for col in df_p1_raw.columns if col.startswith("k_")}

# Apply renaming
df_p1 = df_p1_raw.rename(columns=rename_dict)
df_p1 = df_p1.drop(columns=['smallest_k_matched'])

print(df_p1)
```

```python
# Merge two tables

result = pd.concat([df_p1, df_p2], ignore_index=True)
result.index = range(1, len(result) + 1)
result_rounded = result.round(4)

result_rounded.to_csv("result_table_v2.csv")
result_rounded
```
