```python
import pandas as pd

# Read the CSV file
df_p2 = pd.read_csv('raw_results_p2.csv')

# Extract col1 and col2 from the 'Type' column
df_p2['query_modality'] = df_p2['Type'].str.split('-').str[0]  # First part (e.g., "Spec")
df_p2['document_modality'] = df_p2['Type'].str.split('-').str[-1].str.replace(':', '')  # Last part (e.g., "Spec")
df_p2['embedding_method'] = 'Text Model'  

# Drop original column
df_p2 = df_p2.drop(columns=['Type'])

# Rename
df_p2 = df_p2.replace('Text', 'Text (AltGosling)')
df_p2 = df_p2.replace('LLM', 'Text (AltGosling + LLM)')

# Reorder columns (new columns first, then original)
cols = ['query_modality', 'document_modality', 'embedding_method'] + list(df_p2.columns[:-3])
df_p2 = df_p2[cols]

print(df_p2)

df_p2.to_csv('formatted_p2.csv')
```
